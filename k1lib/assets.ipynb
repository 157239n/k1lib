```python
#export
import k1lib, os, inspect; k1 = k1lib; cli = k1.cli
__all__ = ["k1js_amd", "k1js_umd", "acorn", "acorn_walk"]
```

```python
#export
basePath = os.path.dirname(inspect.getabsfile(k1lib)) + os.sep
@k1.cache(timeout=None)
def readFile(fn):
    with open(f"{basePath}assets{os.sep}{fn}") as f: return f.read()
def k1js_amd(): return readFile("k1js_amd.js")
def k1js_umd(): return readFile("k1js_umd.js")
def acorn(): return readFile("acorn.js")
def acorn_walk(): return readFile("acorn_walk.js")
def chartist_css(): return readFile("chartist.css")
def chartist_js(): return readFile("chartist.js")
def leaflet_css(): return readFile("leaflet.css")
def leaflet_js(): return readFile("leaflet.js")
```
