```python
#export
"""
This is for all short and random quality-of-life utilities."""
import k1lib as k1; import k1lib.cli as cli
from k1lib.web.init import *
bcrypt = k1.dep("bcrypt")
__all__ = ["table", "hashPw", "checkPw", "new", "get", "login"]
```

```python
#export
_table = [None]
def table():
    if _table[0] is None:
        s = db(); s.query("CREATE TABLE IF NOT EXISTS users ( id       INTEGER primary key autoincrement, name     TEXT, username TEXT, password TEXT );")
        s.query("CREATE INDEX IF NOT EXISTS users_username ON users (username);"); _table[0] = s["users"]
    return _table[0]
```

```python
#export
def hashPw(pw:str) -> str: return bcrypt.hashpw(pw.encode(), bcrypt.gensalt()).decode()
def checkPw(pw:str, hashedPw:str) -> bool: return bcrypt.checkpw(pw.encode(), hashedPw.encode())
def new(username:str, password:str):
    if table().lookup(username=username) is not None: raise FailureException(f"Username '{username}' already exists")
    table().insert(username=username, password=password)
def get(idx:int): return table()[idx]
def login(username:str, password:str) -> "User|None":
    user = table().lookup(username=username)
    if user: return user if checkPw(password, user.password) else None
    else: return table().insert(username=username, password=hashPw(password))
```
